```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

filtered = X.filter(
        (pl.col('annee') == 2002) &
        (pl.col('election_type') == 'presidentiel')
    )
numeric_df = filtered.select(features)
corr_matrix = numeric_df.to_pandas().corr(method="pearson")

# Create the heatmap
plt.figure(figsize=(14, 12))

# Mask the upper triangle to avoid redundancy
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

sns.heatmap(
        corr_matrix,
        mask=mask,
        annot=False,       
        fmt=".2f",
        cmap='coolwarm',
        center=0,
        linewidths=0.5,
        cbar_kws={"shrink": .8}
)

plt.title('Correlation Heatmap - Presidential Election 2002', fontsize=15)
plt.tight_layout()
```

```python
import matplotlib.pyplot as plt
import polars as pl


def plot_ts(data, codecommune="75111", features=None):
    if features is None:
        features = ["F_raw_cspcommunes/cadr"]

    cols = ["codecommune", "annee", *features]

    df = (
        data.select(cols)
        .unique(["codecommune", "annee"])
        .filter(pl.col("codecommune") == codecommune)
        .sort("annee")
    )

    pdf = df.to_pandas()

    plt.figure(figsize=(10, 4))
    for feature in features:
        plt.plot(pdf["annee"], pdf[feature], marker="o", label=feature)

    plt.title(f"Time series for codecommune {codecommune}")
    plt.xlabel("Année")
    plt.ylabel("Value")
    plt.grid(True)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
plot_ts(
    data,
    codecommune="18004",
    features=data.select(
        cs.ends_with("pop") & cs.contains("commune") & cs.contains("raw")
    ).columns,
)
```

```python
plot_ts(
    data,
    codecommune="44131",
    features=[
        "F_raw_cspcommunes/pcadr",
        "F_raw_cspcommunes/pchom",
        "F_rank_cspcommunes/pcadr",
        "F_rank_cspcommunes/pchom",
    ],
)
```
